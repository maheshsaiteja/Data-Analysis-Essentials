```python
#nan means a missing value in a data frame we can replace missing values usin

import pandas as pd

data = {
    'Name': ['Ravi', 'Sita', 'None', 'Kiran'],
    'Department': ['CSE', 'None', 'ECE', 'CSE'],
    'marks': [85, 90, 78, 'None']
}

print(data)

# DataFrame
df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Fill NaN values with a string/value
df['Name'] = df['Name'].fillna('Teja')
df['Department'] = df['Department'].fillna('MECH')
df['marks'] = df['marks'].fillna(78)

print("\nAfter filling NaN:")
print(df)
```
