```python
#d.	Adding a new column 
#We can add new column  by dataframe[new column name]=[value1,value2]

import pandas as pd
#creating data
data = {
    "Name": ["Teja", "Ravi", "Sita", "Rahul"],
    "Marks": [85, 65, 90, 55]
}

df = pd.DataFrame(data)

# Displaying the original DataFrame
print("Original DataFrame:")
print(df)

# Adding a new column 
df["  Grade"] = ["A", "B", "A", "C"]

# Displaying the DataFrame after adding the new column
print("\nDataFrame after adding a new column:")
print(df)
```
