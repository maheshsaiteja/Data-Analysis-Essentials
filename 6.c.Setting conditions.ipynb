```python
#c.Setting conditions
# Setting conditions in Pandas means filtering rows from a DataFrame by set condition


import pandas as pd

# Creating a DataFrame
data = {
    "Name": ["Teja", "Ravi", "Sita", "Rahul"],
    "Marks": [85, 65, 90, 55]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Setting condition
print("\nStudents who scored more than 70:")
print(df[df["Marks"] > 70])

```
