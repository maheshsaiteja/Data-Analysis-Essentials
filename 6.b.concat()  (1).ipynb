```python
#b.	concat() 

import pandas as pd

# First DataFrame
df1 = pd.DataFrame({ "Name": ["Teja", "Ajarya"],
    "Age": [18, 19]
})

# Second DataFrame
df2 = pd.DataFrame({
    "Name": ["Atchi", "Mahesh"],
    "Age": [20, 18]
})

# Concatenate both DataFrames
result = pd.concat([df1, df2])

print("First DataFrame:")
print(df1)

print("\nSecond DataFrame:")
print(df2)

print("\nAfter Concatenation:")
print(result)
```
